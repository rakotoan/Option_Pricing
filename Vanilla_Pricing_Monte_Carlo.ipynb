```python
import numpy as np
from scipy.stats import norm
```

```python
# Paramètres
S0 = 100
K = 100
T = 1.0
r = 0.05
sigma = 0.20

N = 100_000
```

```python
# Génération de N variables loi normale centrée réduite

np.random.seed(40) # pour retrouver meme valeur loi qj je rerun le code
Z = np.random.randn(N)

Z
```

```python
# Price at maturity
ST = S0 * np.exp( (r- 0.5 * sigma**2)*T + sigma * np.sqrt(T) * Z)
ST
```

```python
payoff = np.maximum(ST-K,0)  # np.max() donnera valeur max du tableau
payoff
```

```python
# Call price
price = np.mean(payoff)*np.exp(-r*T)
price
```

```python
# Price Error

# ecart-type empirique payoff
ecart_type_payoff = np.std(payoff)
ecart_type_payoff
```

```python
# Error
error = ecart_type_payoff * (1/np.sqrt(N)) * np.exp(-r*T)
error
```

```python
# IC 95%

b = price - 1.96*error
h = price + 1.96*error

IC = [b,h]
IC
```
